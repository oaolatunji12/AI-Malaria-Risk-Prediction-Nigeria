```python
def build_dictionary(meta, variables, dataset_name):

    rows = []

    for var in variables:

        label = meta.column_names_to_labels.get(var)

        label_set = meta.variable_to_label.get(var)

        codes = meta.value_labels.get(label_set)

        rows.append({
            "Dataset": dataset_name,
            "Variable": var,
            "Label": label,
            "Code_Definitions": str(codes),
            "Notes": ""
        })

    return pd.DataFrame(rows)
```

```python
dict_A = build_dictionary(
    meta2021,
    harmonized_vars[1:],  # remove survey_year
    "Dataset A"
)

dict_B = build_dictionary(
    meta2015,
    treatment_vars,
    "Dataset B"
)

dict_C = build_dictionary(
    meta2021,
    awareness_vars,
    "Dataset C"
)
```
