# To start cleaning Dataset D

```python
# Load dataset D
import pandas as pd
import numpy as np

Dataset_D = pd.read_excel("Dataset_D_Final2.xlsx")

print(Dataset_D.shape)
Dataset_D.head()
```

```python
# Fixing their data types
# Categories
cat_cols = [
    'HV024',
    'HV025',
    'HV270',
    'HV219',
    'survey_year'
]

Dataset_D[cat_cols] = (
    Dataset_D[cat_cols]
    .astype('category')
)

# Binary
binary_cols = [
    'slept_under_treated_net',
    'slept_under_llin',
    'malaria_positive',
    'HV206',
    'HV207',
    'HV208',
    'HV209',
    'pfalciparum',
    'pmalariae'
]

Dataset_D[binary_cols] = (
    Dataset_D[binary_cols]
    .astype('int8')
)

# Checking data types
Dataset_D.info()
```

```python
# Checking for missing
missing_percent = (
    Dataset_D
    .isnull()
    .mean()
    .mul(100)
    .round(2)
)

missing_percent.sort_values(
    ascending=False
)
```

# Handling the missing

```python
# To handle the missing values in binary variables, fill them with 0
binary_cols = [
    'HV206',
    'HV207',
    'HV208',
    'HV209',
    'slept_under_treated_net',
    'slept_under_llin',
    'pfalciparum',
    'pmalariae'
]

for col in binary_cols:
    Dataset_D[col] = (
        Dataset_D[col]
        .fillna(0)
    )


# To handle the missing values in continous variables, fill them with the median

continuous = [
    'age',
    'HV220',
    'HV216',
    'HV271',
    'HML1'
]

for col in continuous:
    Dataset_D[col] = (
        Dataset_D[col]
        .fillna(
            Dataset_D[col].median()
        )
    )

# To handle Rapid Test and Blood Smear Variables Because these tests were not conducted for everybody, I would not impute them with medians.
# but instead fill it with -1 and this mean "NOT TESTED"

Dataset_D['blood_smear_result'] = (
    Dataset_D['blood_smear_result']
    .fillna(-1)
)

Dataset_D['rapid_test_result'] = (
    Dataset_D['rapid_test_result']
    .fillna(-1)
)

# To fill 1 and 2 coding bianry
for col in ['HV201','HV205','HV226']:
    Dataset_D[col] = Dataset_D[col].fillna(
        Dataset_D[col].mode()[0]
    )


# To finally check if all missing has been handled so everything becomes ZERO
Dataset_D.isnull().sum().sort_values(
    ascending=False
)
```
